```python
#!pip install hmmlearn

import numpy as np
from hmmlearn import hmm

states = ["1", "2", "3", "4"]
observations = ["red", "green", "blue"]
start_prob = np.array([0.25, 0.25, 0.25, 0.25])

trans_mat = np.array([
    [0.2, 0.4, 0, 0.4],
    [0.4, 0.2, 0.4, 0],
    [0, 0.4, 0.2, 0.4],
    [0.4, 0, 0.4, 0.2]
])

emission_mat = np.array([
    [0.7, 0.2, 0.1],
    [0.2, 0.6, 0.2],
    [0.1, 0.3, 0.6],
    [0.5, 0.2, 0.3]
])

model = hmm.CategoricalHMM(n_components=4, init_params="")
model.startprob_ = start_prob
model.transmat_ = trans_mat
model.emissionprob_ = emission_mat

print("Start probabilities:\n", model.startprob_)
print("\nTransition matrix:\n", model.transmat_)
print("\nEmission matrix:\n", model.emissionprob_)

observatii_map = {"red": 0, "green": 1, "blue": 2}
camere_map_invers = {0: "1", 1: "2", 2: "3", 3: "4"}

observatii = ["red", "green", "blue", "blue", "red"]
obs_seq = np.array([[observatii_map[n]] for n in observatii])

logprob, stari_idx = model.decode(obs_seq, algorithm="viterbi")

secventa_camere = [camere_map_invers[i] for i in stari_idx]
probabilitate = np.exp(logprob)

print("Cea mai probabilă secvență de camere:")
print(secventa_camere)
print(f"\nLog-probabilitate: {logprob:.6f}")
print(f"Probabilitate exactă: {probabilitate:.6e}")
```
